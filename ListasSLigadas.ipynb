```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None

class LSL:
    def __init__(self):
        self.cabecera = None

    def insertar(self, valor):
        nuevo_nodo = Nodo(valor) # Instanciación del objeto Nodo
        if self.cabecera == None: # Primer valor
            self.cabecera = nuevo_nodo
        else:
            nodo_actual = self.cabecera # Puntero
            while nodo_actual.siguiente: # mientras sea diferente de None
                nodo_actual = nodo_actual.siguiente
            nodo_actual.siguiente = nuevo_nodo # inserta el valor

    def imprimir(self):
        if self.cabecera is None:
            print("LSL está vacía")
            print(None)
        else:
            nodo_actual = self.cabecera # Puntero
            while nodo_actual:
                print(nodo_actual.valor, end = "->")
                nodo_actual = nodo_actual.siguiente
            print(None)

    def eliminar(self):
        if self.cabecera is None:
            print("LSL está vacía")
            print(None)
        else:
            self.cabecera = self.cabecera.siguiente
            nodo_actual = self.cabecera

    def buscar(self, x_valor):
        nodo_actual = self.cabecera
        pos = 0
        while nodo_actual:
            if nodo_actual.valor == x_valor:
                print("Valor encontrado en posición {:.0f}".format(pos))
                return True
            nodo_actual = nodo_actual.siguiente
            pos += 1
        return False

    def unirListas(self, Lista2):
        if self.cabecera == None:
            print("LSL está vacía")
            return False
        nodo_actual = self.cabecera # Puntero
        while nodo_actual.siguiente: # mientras sea diferente de None
            nodo_actual = nodo_actual.siguiente
        nodo_actual.siguiente = Lista2.cabecera
```

```python
Lista1 = LSL()
Lista1.insertar(2)
Lista1.insertar(5)
Lista1.insertar(10)
Lista1.imprimir()

Lista2 = LSL()
Lista2.insertar(7)
Lista2.insertar(8)
Lista2.insertar(15)
Lista2.imprimir()

Lista1.unirListas(Lista2)
Lista1.imprimir()
```
