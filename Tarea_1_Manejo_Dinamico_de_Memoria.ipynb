### Ejercicio 1: Concatenación de LSL

**Descripción:** Dado dos LSL, *Lista_1* y *Lista_2*, diseñe un algoritmo que permita concatenar ambas listas en una nueva LSL. Posteriormente, la nueva LSL debe estar ordenada de forma ascendente para retornar al usuario la LSL final.

**Idea:** Si $Lista_1 = 5 \rightarrow 7 \rightarrow 8 \rightarrow NULL$ y $Lista_2 = 7 \rightarrow 5 \rightarrow 2 \rightarrow NULL$, la nueva LSL, $NuevaLSL = 2 → 5 → 5 \rightarrow 7 → 7 \rightarrow 8 → NULL$

```python
# Diseño del algoritmo
# //
class Nodo:
  def __init__(self, valor):
    self.valor = valor
    self.siguiente = None

class LSL:
  def __init__(self):
    self.cabecera = None

  def insertar(self, valor):
    nuevo_nodo = Nodo(valor) # Instanciación del objeto Nodo
    if self.cabecera == None: # Primer valor
      self.cabecera = nuevo_nodo
    else:
      nodo_actual = self.cabecera # Puntero
      while nodo_actual.siguiente: # mientras sea diferente de None
        nodo_actual = nodo_actual.siguiente
      nodo_actual.siguiente = nuevo_nodo # inserta el valor

  def imprimir(self):
    if self.cabecera is None:
      print("LSL está vacía")
      print(None)
    else:
      nodo_actual = self.cabecera # Puntero
      while nodo_actual:
        print(nodo_actual.valor, end = "->")
        nodo_actual = nodo_actual.siguiente
      print(None)

  def eliminar(self):
    if self.cabecera is None:
      print("LSL está vacía")
      print(None)
    else:
      self.cabecera = self.cabecera.siguiente
      nodo_actual = self.cabecera

  def buscar(self, x_valor):
    nodo_actual = self.cabecera
    pos = 0
    while nodo_actual:
      if nodo_actual.valor == x_valor:
        print("Valor encontrado en posición {:.0f}".format(pos))
        return True
      nodo_actual = nodo_actual.siguiente
      pos += 1
    return False

  def unirListas(self, Lista2):
    if self.cabecera == None:
      print("LSL está vacía")
      return False
    nodo_actual = self.cabecera # Puntero
    while nodo_actual.siguiente: # mientras sea diferente de None
      nodo_actual = nodo_actual.siguiente
    nodo_actual.siguiente = Lista2.cabecera
    Lista2.cabecera = None
    return

  def ordenar(self):
    if self.cabecera is None:
      print("LSL está vacía")
      return False
    i = 0
    n = 0
    nodo_actual = self.cabecera
    while nodo_actual:
      nodo_actual = nodo_actual.siguiente
      n += 1
    while i <= n:
      nodo_actual = self.cabecera.siguiente
      nodo_anterior = self.cabecera
      while nodo_actual:
        if nodo_anterior.valor > nodo_actual.valor:
          nodo_actual.valor, nodo_anterior.valor = nodo_anterior.valor, nodo_actual.valor
        nodo_actual = nodo_actual.siguiente
        nodo_anterior = nodo_anterior.siguiente
      i += 1

Lista1 = LSL()
Lista1.insertar(5)
Lista1.insertar(7)
Lista1.insertar(8)
Lista1.imprimir()
Lista2 = LSL()
Lista2.insertar(7)
Lista2.insertar(5)
Lista2.insertar(2)
Lista2.imprimir()
Lista1.unirListas(Lista2)
Lista1.ordenar()
Lista1.imprimir()
# //
```
